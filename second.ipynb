```python
import matplotlib.pyplot as plt
```

```python
area = [
    500, 550, 600, 650, 700, 750, 800, 850, 900, 950,
    1000, 1050, 1100, 1150, 1200, 1250, 1300, 1350, 1400, 1450,
    1500, 1550, 1600, 1650, 1700, 1750, 1800, 1850, 1900, 1950,
    2000, 2050, 2100, 2150, 2200, 2250, 2300, 2350, 2400, 2450,
    2500, 2550, 2600, 2650, 2700, 2750, 2800, 2850, 2900, 3000
]

price = [
    45, 72, 54, 81, 63, 90, 70, 105, 76, 98,
    112, 89, 128, 101, 139, 115, 146, 122, 158, 131,
    170, 136, 181, 148, 193, 157, 205, 164, 217, 176,
    225, 188, 241, 195, 256, 207, 269, 218, 278, 232,
    290, 243, 307, 251, 315, 266, 330, 274, 345, 292
]
```

```python
m = 0
c = 0
total = len(area)
learningRate = 0.001
losses = list()
```

normalizing area

```python
meanArea = sum(i for i in area) / len(area)
std_area = (sum((x - meanArea)**2 for x in area) / len(area)) ** 0.5
areaNormal = [(x - meanArea) / std_area for x in area]
```

```python
plt.plot(areaNormal, price, marker = ".", linestyle = "None")

predictedValue = [(m * i) + c for i in area]
plt.plot(areaNormal, predictedValue, marker = ".")
```

```python
def loss():
    return sum((i - j)**2 for i, j in zip(predictedValue, price)) / (2 * total)
```

```python
for _ in range(1000):
    dm = sum((i - j)*k for i, j, k in zip(predictedValue, price, areaNormal)) * (2 / total)
    dc = sum((i - j) for i, j in zip(predictedValue, price)) * (2 / total)

    m = m - learningRate * dm
    c = c - learningRate * dc

    predictedValue = [(m * i) + c for i in areaNormal]

lossValue = loss()
losses.append((lossValue, m, c))

plt.plot(areaNormal, price, marker = ".", linestyle = "None")
plt.plot(areaNormal, predictedValue, marker = ".")
```
