```python
x = [2000, 2001, 2002, 2003]
y1 = [45, 63, 100, 323]
y2 = [30, 120, 200, 173]

about = dict(marker = ".", markersize = "20", markerfacecolor = "black",
         markeredgecolor = "white", linestyle = "dashed", linewidth = "1",
                  color = "green")

plt.plot(x, y1, **about)

plt.plot(x, y2, marker = ".", ms = "20", linestyle = "dashed", color = "orange",
         markeredgecolor = "white", markerfacecolor = "black")

plt.title("size of the Dick", fontsize = "13")
plt.xlabel("year")
plt.ylabel("size")
plt.tick_params(axis = "both", colors = "red")
plt.xticks(x)

plt.show()
```

```python
x = [2000, 2001, 2002, 2003]
y1 = [45, 63, 100, 323]
y2 = [30, 120, 200, 173]

about = dict(marker = ".", markersize = "20", markerfacecolor = "black",
         markeredgecolor = "white", linestyle = "dashed", linewidth = "1",
                           color = "green")

plt.plot(x, y1, **about)

plt.plot(x, y2, marker = ".", ms = "20", linestyle = "dashed", color = "orange",
          markeredgecolor = "white", markerfacecolor = "black")

plt.grid(axis = "y", linewidth = 2)
plt.title("size of the Dick", fontsize = "13")
plt.xlabel("year")
plt.ylabel("size")
plt.tick_params(axis = "both", colors = "red")
plt.xticks(x)

plt.show()
```
