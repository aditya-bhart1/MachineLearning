```python
from coffeeData import x, y
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential([
    tf.keras.Input(shape = (2,)),       #describes the shape of one example
    Dense(units = 3, activation = "sigmoid", name = "one"),
    Dense(units = 1, activation = "sigmoid", name = "two")
])

model.summary()
```

```python
W1, b1 = model.get_layer("one").get_weights()
W2, b2 = model.get_layer("two").get_weights()

print(f"W1{W1.shape}:\n", W1, f"\nb1{b1.shape}:", b1)
print()
print(f"W2{W2.shape}:\n", W2, f"\nb2{b2.shape}:", b2)
```

```python
model.compile(
    loss = tf.keras.losses.BinaryCrossentropy(),
    optimizer = tf.keras.optimizers.Adam(learning_rate = 0.01)
)

history = model.fit(
    x, y,
    epochs = 10
)
```

```python
history.history["loss"]
```

```python
model.evaluate(x, y)
```
