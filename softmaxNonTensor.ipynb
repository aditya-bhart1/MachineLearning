```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("digits.csv")
y = df["label"].values
x = df.drop("label", axis = 1).values

y[:6]   #actual digits
```

25 -> 15 -> 10

```python
def relu(x):
    return np.maximum(0, x)
```

```python
def softmax(z):     # ai code, i only know the formula not how to put it into code hehe
    exp_z = np.exp(z)
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)
```

```python
W1 = np.random.randn(400, 25) * 0.01
b1 = np.zeros(25)

W2 = np.random.randn(25, 15) * 0.01
b2 = np.zeros(15)

W3 = np.random.randn(15, 10) * 0.01
b3 = np.zeros(10)
```

```python
Z1 = x @ W1 + b1
A1 = relu(Z1)

Z2 = A1 @ W2 + b2
A2 = relu(Z2)

Z3 = A2 @ W3 + b3
A3 = softmax(Z3)
```

accuracy without back propagation

```python
predictions = np.argmax(A3, axis=1)
correct = 0

for i in range(5000):
    if predictions[i] == y[i]:
        correct += 1

print(f"correct: {correct}, total: {len(y)}")
print(f"accuracy: {correct/len(y) * 100}%")
```
