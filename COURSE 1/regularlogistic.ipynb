```python
import numpy as np
import matplotlib.pyplot as plt
from math import log
from math import e
```

```python
data = np.loadtxt("logisticDataset.csv", delimiter=",", skiprows=1, ndmin=2)

studied = data[:, 0]
slept = data[:, 1]
result = data[:, 2]

x = np.column_stack((studied, slept))
```

```python
greek = 1
learningRate = 0.01
total = len(result)
c = 0
m = np.zeros(2)
dm = np.zeros(2)
costs = list()
```

```python
def sigmoid(y):
    return 1/(1 + e**(-y))
```

```python
for i in range(1000000):
    y = np.dot(x, m) + c
    prediction = sigmoid(y)
    loss = prediction - result

    dm = (np.dot(x.T, loss) + greek * m) / total
    dc = (np.sum(loss)) / total

    m = m - learningRate * dm
    c = c - learningRate * dc
```
