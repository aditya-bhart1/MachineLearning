```python
mean = x.mean(axis=0)
std = x.std(axis=0)
x_norm = (x - mean) / std
x = x_norm
```

```python
a = [int(x[i][0]) for i in range(50)]
b = [int(y[i]) for i in range(50)]

plt.plot(a, b, marker = "x", linestyle = "None")
```

```python
def loss(m, c, x, y):
    prediction = np.dot(x, m) + c
    error = prediction - y
    return int(np.mean(error**2))
```

```python
m = np.zeros(4)
dm = np.zeros(4)
learningRate = 0.01
losses = list()
c = 0
ms = list()
cs = list()
iterations = list()
```

```python
for k in range(5000):
    prediction = np.dot(x, m) + c
    error = prediction - y

    for i in range(4):
        dm[i] = ((np.sum(np.dot(error, x[:, i]))) * 2) / 50
    dc = ((np.sum(error)) * 2) / 50

    for i in range(4):
        m[i] = m[i] - learningRate * dm[i]
    c = c - learningRate * dc

    if k%100 == 0:
        lossValue = loss(m, c, x, y)
        losses.append(lossValue)
        ms.append((m))
        cs.append(c)
        iterations.append(k)
```

```python
print(f"m: {ms}, c: {cs}")
print(losses[::-1])
```

```python
# loss vs iterations
plt.plot(iterations, losses)
plt.title("learning curve")
plt.show()
```
