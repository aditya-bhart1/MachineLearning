```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = 0
c = 0
learningRate = 0.001
```

```python
area = [650, 785, 1200, 1450, 1600, 1850, 2100, 2350, 2600, 3000]
price = [70, 85, 145, 165, 175, 200, 225, 240, 260, 300]
total = len(area)

plt.plot(area, price, marker = ".", markersize = "10", linestyle = "None")

# standard deviation for area
mean_area = sum(area)/len(area)
std_area = (sum((x - mean_area)**2 for x in area) / len(area)) ** 0.5
area_norm = [(x - mean_area) / std_area for x in area]  
print(area_norm)

# y = m * x + c
predictedValue = [(m*i + c) for i in area]
plt.plot(area, predictedValue, marker = ".")
```

```python
dm = sum((i - j)*k for i,j,k in zip(predictedValue, price, area_norm)) * (2 / total)
dc = sum((i - j) for i, j in zip(predictedValue, price)) * (2 / total)

print(f"dm: {dm}, dc: {dc}")
```

```python
for _ in range(4000):
    dm = sum((i - j)*k for i,j,k in zip(predictedValue, price, area_norm)) * (2 / total)
    dc = sum((i - j) for i, j in zip(predictedValue, price)) * (2 / total)

    m = m - learningRate * dm
    c = c - learningRate * dc

    predictedValue = [(m*i + c) for i in area_norm]

plt.plot(area_norm, predictedValue, marker = ".")
plt.plot(area_norm, price, marker = ".", markersize = "10", linestyle = "None")

print(f"m: {m}, c: {c}")
print(f"dm: {dm}, dc: {dc}")
```
